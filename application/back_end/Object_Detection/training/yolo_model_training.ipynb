```python
from ultralytics import YOLO
from pathlib import Path

yaml_path = Path(r"T:\AI\Fruit_model_traininig\fruit_yolo_dataset\dataset.yaml")

if not yaml_path.exists():
    raise FileNotFoundError(f"Could not find: {yaml_path}")

model = YOLO("yolov8s.pt")

model.train(
    data=str(yaml_path),
    epochs=75,
    imgsz=640,
    batch=16,
    device=0,
    workers=4,
    cache=False,
    patience=15,
    project=r"T:\AI\Fruit_model_traininig\training_runs",  # ✅ custom base folder
    name="yolov8s_fruit_v1",                                # ✅ run name
)
```
